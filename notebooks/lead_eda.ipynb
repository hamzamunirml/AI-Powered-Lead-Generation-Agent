Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load Dataset

```python
df = pd.read_csv('../data/ml_ready_dataset.csv')
```

```python
# 1. First 5 records
print("First 5 records:")
print(df.head())
```

```python
# 2. Total companies
print(f"\nTotal companies: {len(df)}")
```

```python
# 3. By country
print("\nBy country:")
print(df['Country'].value_counts())
```

```python
# 4. By industry
if 'Industry' in df.columns:
    print("\nBy industry:")
    print(df['Industry'].value_counts())
```

```python
# 5. Lead quality
if 'lead_quality_score' in df.columns:
    print("\nLead Quality:")
    quality_map = {3:'High',2:'Medium',1:'Low'}
    df['Quality'] = df['lead_quality_score'].map(quality_map)
    print(df['Quality'].value_counts())
```

```python
# 6. Average services
print(f"\nAverage services: {df['services_count'].mean():.2f}")
```

```python
# 7. Contact forms percentage
print(f"Contact forms: {(df['contact_form'].sum()/len(df))*100:.1f}%")
```

```python
plt.figure(figsize=(10,6))
df['Country'].value_counts().plot(kind='bar', color=['#FF6B6B','#4ECDC4','#45B7D1'])
plt.title('Companies by Country', fontweight='bold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../docs/country_distribution.png')
plt.show()
```

```python
# Chart 2: Lead quality distribution
plt.figure(figsize=(10,6))
df['lead_quality_score'].value_counts().sort_index().plot(kind='bar', 
    color=['#2ECC71','#F39C12','#E74C3C'])
plt.title('Lead Quality Distribution', fontweight='bold')
plt.xlabel('Quality (3=High, 2=Medium, 1=Low)')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('../docs/lead_quality_distribution.png')
plt.show()
```

```python
# Chart 3: Services distribution
plt.figure(figsize=(10,6))
df['services_count'].value_counts().sort_index().plot(kind='bar', color='#3498DB')
plt.title('Services per Company', fontweight='bold')
plt.xlabel('Number of Services')
plt.ylabel('Number of Companies')
plt.tight_layout()
plt.savefig('../docs/services_distribution.png')
plt.show()
```
