## Import Libraries 

```python
import pandas as pd
import joblib
import shap
import matplotlib.pyplot as plt
```

## Load model and dataset

```python
# Load model and data
model = joblib.load('notebooks/saved_models/best_lead_model.pkl')
df = pd.read_csv('data/scored_leads.csv')
```

# Features (X) and Target (y)


```python
features = ['website_exists', 'contact_form', 'services_count', 'country_score']
X = df[features]
y = df['Predicted Lead Class']
```

## Display SHAP

```python
# Create SHAP explainer
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X)
```

```python
# Feature importance plot
plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X, feature_names=features, show=False)
plt.tight_layout()
plt.savefig('feature_importance.png')
plt.show()
```

```python
# Feature importance bar plot
shap.summary_plot(shap_values, X, feature_names=features, plot_type="bar", show=False)
plt.tight_layout()
plt.savefig('feature_importance_bar.png')
plt.show()

print("✅ SHAP analysis complete!")
print("Saved: feature_importance.png, feature_importance_bar.png")
```
